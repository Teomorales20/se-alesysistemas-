#**Sistema de Detección de Madurez de Frutos usando Espectroscopía Eléctrica**

```python
# 📦 Instala librerías necesarias
!pip install matplotlib pandas

# 📚 Importa librerías
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display, clear_output
import time

# 🔧 Configuración de simulación
tiempo = np.linspace(0, 10, 100)  # Tiempo simulado de 10 segundos
madurez = []                      # Lista para guardar niveles de madurez simulados

# Función que simula lectura del sensor espectroscópico
def leer_sensor_espectroscopico(potenciometro_valor):
    # Simula un valor de espectro basado en el potenciómetro
    ruido = np.random.normal(0, 0.5)
    return 2.5 * potenciometro_valor + ruido

# Función para determinar nivel de madurez según espectroscopía
def calcular_madurez(valor_sensor):
    if valor_sensor < 2:
        return "Verde"
    elif 2 <= valor_sensor < 4:
        return "En Maduración"
    else:
        return "Maduro"

# 🔁 Simulación en vivo
for t in tiempo:
    clear_output(wait=True)

    # Simula valor del potenciómetro (entre 0 y 1)
    pot_valor = np.random.rand()

    # Simula lectura del sensor
    valor_sensor = leer_sensor_espectroscopico(pot_valor)

    # Clasifica madurez
    estado = calcular_madurez(valor_sensor)
    madurez.append((t, pot_valor, valor_sensor, estado))

    # Muestra datos simulados
    print(f"⏱ Tiempo: {t:.2f}s")
    print(f"🔘 Potenciómetro: {pot_valor:.2f}")
    print(f"🔬 Sensor Espectroscopía: {valor_sensor:.2f} V")
    print(f"🍎 Estado del fruto: {estado}")

    # Espera para animación
    time.sleep(0.1)

# 📊 Mostrar resultados
df = pd.DataFrame(madurez, columns=["Tiempo (s)", "Potenciómetro", "Sensor (V)", "Madurez"])
df["Color"] = df["Madurez"].map({"Verde": "green", "En Maduración": "orange", "Maduro": "red"})

plt.figure(figsize=(10,5))
plt.scatter(df["Tiempo (s)"], df["Sensor (V)"], c=df["Color"], label=df["Madurez"])
plt.plot(df["Tiempo (s)"], df["Sensor (V)"], alpha=0.3, color='gray')
plt.title("Detección de Madurez del Fruto (Simulación)")
plt.xlabel("Tiempo (s)")
plt.ylabel("Voltaje del Sensor")
plt.grid(True)
plt.show()
```
