**1. Transformada Rápida de Fourier (FFT) a una Señal Sintética
Código base en Python (Colab o Streamlit):**

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.fft import fft, fftfreq

# Señal sintética: suma de dos senoides
fs = 1000  # frecuencia de muestreo
T = 1.0 / fs
t = np.arange(0.0, 1.0, T)
f1, f2 = 50, 150  # frecuencias componentes

# Señal: suma de dos senos
x = np.sin(2*np.pi*f1*t) + 0.5*np.sin(2*np.pi*f2*t)

# Transformada de Fourier
N = len(x)
yf = fft(x)
xf = fftfreq(N, T)[:N//2]

# Visualización
plt.figure(figsize=(10,4))
plt.plot(xf, 2.0/N * np.abs(yf[0:N//2]))
plt.title('Espectro de Magnitud')
plt.xlabel('Frecuencia (Hz)')
plt.ylabel('Magnitud')
plt.grid()
plt.show()
```

 **2. Filtro Paso Bajo Digital FIR (Diseño y Aplicación)**

```python
from scipy.signal import firwin, lfilter

# Diseño del filtro
nyq = fs / 2
cutoff = 100  # frecuencia de corte
numtaps = 101
b = firwin(numtaps, cutoff / nyq)

# Aplicación del filtro
x_filtrada = lfilter(b, 1.0, x)

# Gráfica de la señal filtrada
plt.figure(figsize=(10,4))
plt.plot(t, x, label='Original')
plt.plot(t, x_filtrada, label='Filtrada', alpha=0.7)
plt.title('Señal Filtrada con Filtro Paso Bajo FIR')
plt.xlabel('Tiempo (s)')
plt.legend()
plt.grid()
plt.show()
```

**3. Transformada de Hilbert y Señal Analítica**

```python
from scipy.signal import hilbert

# Transformada de Hilbert
x_analytic = hilbert(x)
x_I = np.real(x_analytic)
x_Q = np.imag(x_analytic)

# Gráfica de I y Q
plt.figure(figsize=(10,4))
plt.plot(t, x_I, label='I (original)')
plt.plot(t, x_Q, label='Q (Hilbert)', linestyle='--')
plt.title('Componentes I/Q de la señal')
plt.legend()
plt.grid()
plt.show()
```
