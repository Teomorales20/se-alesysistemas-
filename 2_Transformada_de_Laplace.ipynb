```python
import numpy as np
import matplotlib.pyplot as plt

def sistema(x):
    N = len(x)
    y = np.zeros(N)
    for n in range(1, N):  # desde 1 para que y[n-1] exista
        y[n] = x[n]/3 + 2*x[n-1] - y[n-1]
    return y

# Señales de prueba
n = np.arange(0, 20)
x1 = np.sin(0.2 * np.pi * n)
x2 = np.cos(0.2 * np.pi * n)
x_sum = x1 + x2

# Respuestas individuales y suma
y1 = sistema(x1)
y2 = sistema(x2)
y_sum = sistema(x_sum)

# Comprobación de linealidad
plt.figure(figsize=(10, 6))
plt.plot(n, y_sum, label='y(x1 + x2)')
plt.plot(n, y1 + y2, '--', label='y(x1) + y(x2)')
plt.title('Comprobación de linealidad')
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Definición del sistema
def sistema(x):
    y = np.cumsum(x**2)  # Acumulación de x[k]^2
    return y

# Vector de tiempo discreto
n = np.arange(0, 20)

# Señal de entrada (puedes cambiarla por cualquier otra)
x = np.sin(0.2 * np.pi * n)

# Aplicar el sistema
y = sistema(x)

# Graficar la salida
plt.stem(n, y, basefmt=" ", linefmt='b-', markerfmt='bo')
plt.title("Salida del sistema: $y[n] = \sum_{k=-\infty}^{n} x^2[k]$")
plt.xlabel("n")
plt.ylabel("y[n]")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.signal import medfilt

# Señal de entrada
n = np.arange(0, 20)
x = np.sin(0.2 * np.pi * n) + 0.3 * np.random.randn(len(n))  # señal con ruido

# Aplicar mediana con ventana de 3 muestras
y = medfilt(x, kernel_size=3)

# Graficar
plt.figure(figsize=(10, 4))
plt.stem(n, x, linefmt='gray', markerfmt='go', basefmt=' ')
plt.plot(n, y, 'r-o', label='y[n] = mediana(x[n-1], x[n], x[n+1])')
plt.title("Filtro de Mediana (ventana de tamaño 3)")
plt.xlabel("n")
plt.legend()
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Parámetros del sistema
A = 2.0  # Ganancia
B = 1.5  # Término constante

# Tiempo continuo simulado
t = np.linspace(0, 2 * np.pi, 1000)

# Señal de entrada
x = np.sin(t)

# Aplicación del sistema: y(t) = A * x(t) + B
y = A * x + B

# Graficar
plt.figure(figsize=(10, 4))
plt.plot(t, x, label='x(t) = sin(t)', linestyle='--')
plt.plot(t, y, label=f'y(t) = {A}·x(t) + {B}', linewidth=2)
plt.title('Sistema: y(t) = A·x(t) + B')
plt.xlabel('t')
plt.ylabel('Amplitud')
plt.grid(True)
plt.legend()
plt.tight_layout()
plt.show()
```
