**Solución: Función de Transferencia en Lazo Abierto para el Sistema Masa-Resorte-Amortiguador**

La ecuación diferencial que describe el movimiento del sistema masa-resorte-amortiguador está dada por la segunda ley de Newton:

$$F(t) = m \frac{d^2x}{dt^2} + b \frac{dx}{dt} + kx(t)$$

Donde:

$F(t)$ es la fuerza externa aplicada al sistema.
- $m$ es la masa.
- $b$ es el coeficiente de amortiguación.
- $k$ es la constante del resorte.
- $x(t)$ es el desplazamiento de la masa.

Para encontrar la función de transferencia, aplicamos la Transformada de Laplace a ambos lados de la ecuación, asumiendo condiciones iniciales cero ($x(0) = 0$, $x'(0) = 0$):

$$\mathcal{L}\{F(t)\} = \mathcal{L}\{m \frac{d^2x}{dt^2} + b \frac{dx}{dt} + kx(t)\}$$

$$F(s) = m \mathcal{L}\{\frac{d^2x}{dt^2}\} + b \mathcal{L}\{\frac{dx}{dt}\} + k \mathcal{L}\{x(t)\}$$

Usando las propiedades de la Transformada de Laplace para derivadas ( $\mathcal{L}\{\frac{dx}{dt}\} = sX(s)$ y $\mathcal{L}\{\frac{d^2x}{dt^2}\} = s^2X(s)$ con condiciones iniciales cero ) :

$$F(s) = m s^2 X(s) + b s X(s) + k X(s)$$

Ahora, factorizamos $X(s)$:

$$F(s) = X(s) (m s^2 + b s + k)$$

La función de transferencia en lazo abierto, $G(s)$, se define como la relación entre la salida $X(s)$ y la entrada $F(s)$:

$$G(s) = \frac{X(s)}{F(s)}$$

Despejando $\frac{X(s)}{F(s)}$ de la ecuación anterior:

$$G(s) = \frac{1}{m s^2 + b s + k}$$

Esta es la función de transferencia en lazo abierto para el sistema masa-resorte-amortiguador.
